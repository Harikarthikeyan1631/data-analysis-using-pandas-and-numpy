```python
import numpy as np
import pandas as pd
```

```python
# Week 1 temperature data
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Display the array
print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

# Maximum temperature
print("Maximum Temperature:", np.max(temperatures_w1))

# Minimum temperature
print("Minimum Temperature:", np.min(temperatures_w1))

# Average temperature
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
# First three days
print("First Three Days:")
print(temperatures_w1[:3])

# Weekend temperatures (last two days)
print("Weekend Temperatures:")
print(temperatures_w1[-2:])

# Middle three days
print("Middle Three Days:")
print(temperatures_w1[2:5])
```

```python
# Creating a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# Display the Series
print("Student Marks:")
print(marks)
```

```python
print("Mark of 1st Rank Student:")
print(marks.iloc[0])
```

```python
print("Top Three Rank Students:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
print("Mark of 3rd Rank Student:")
print(marks.iloc[2])
```

```python
print("Students Scoring More Than 90:")

high_marks = marks[marks > 90]

print(high_marks)
```

```python
# Modify Rank1 marks to 100
marks.loc['Rank1'] = 100

print("Updated Student Marks:")
print(marks)
```

```python
# Remove Rank5 from the Series
marks = marks.drop('Rank4')

print("Series After Removing Last Rank:")
print(marks)
```

```python
# Calculate CGPA
cgpa = marks / 10

print("Student CGPA:")
print(cgpa)
```

```python
# Creating a Pandas DataFrame

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Create DataFrame
transactions = pd.DataFrame(data)

# Display DataFrame
print("Transaction Dataset:")
print(transactions)
```

```python
print("First Five Records:")
print(transactions.head())
```

```python
print("Last Five Records:")
print(transactions.tail())
```

```python
print("DataFrame Shape:")
print(transactions.shape)
```

```python
print("Column Names:")
print(transactions.columns)
```

```python
print("DataFrame Information:")
transactions.info()
```

```python
print("Product Category and Amount:")
print(transactions[['ProductCategory', 'Amount']])
```

```python
print("Last Three Columns:")
print(transactions.iloc[:, -3:])
```

```python
print("North Region Transactions Above 200:")

filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)
```

```python
print("Product Category Counts:")

print(transactions['ProductCategory'].value_counts())
```

```python
print("Unique Regions:")

print(transactions['Region'].unique())
```

```python
print("Average Transaction Amount by Region:")

region_average = transactions.groupby('Region')['Amount'].mean()

print(region_average)
```

```python
#MANIPULATING THE DATAFRAME
```

```python
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print("Updated Transaction Dataset:")
print(transactions)
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print("Dataset After Adding Discount:")
print(transactions)
```

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

print("Dataset After Removing Transaction 109:")
print(transactions)
```

```python
print(transactions.columns)
```

```python
# Delete Discount column only if it exists

if 'Discount' in transactions.columns:
    transactions = transactions.drop(columns=['Discount'])
    print("Discount column deleted successfully.")
else:
    print("Discount column is already deleted.")

print("\nFinal Transaction Dataset:")
print(transactions)
```
